# GUI 실습

## ipywidgets

**ipywidgets**는 Jupyter Notebook이나 Colab 같은 노트북 환경에서 버튼, 입력창, 슬라이더, 출력창 같은 위젯을 만들고 이벤트를 처리할 수 있게 해 주는 인터랙티브 GUI 프로그램 작성 도구

- GUI 프로그래밍: 화면에 보이는 위젯을 배치하고, 사용자의 클릭/입력 같은 이벤트에 따라 핸들러 함수가 실행되도록 연결하는 프로그래밍 방식
    - 위젯 : 버튼, 입력창, 출력창, 체크박스 등 화면 구성 요소
    - 배치 : 위젯을 화면에 보여 주는 것
    - 이벤트 : 클릭, 입력, 선택 등 사용자의 동작
    - 핸들러 : 이벤트가 발생했을 때, 실행되는 함수

- ipywidgets.Output  

    Colab/Jupyter 화면의 특정 영역에서 GUI 프로그램의 결과 표시창 역할을 하는 위젯

````python
import ipywidgets as widgets
from IPython.display import display

output = widgets.Output()

with output:
    print("Hello, World!")

# 위젯 객체를 표시 합니다.
display(output)
````

위젯 객체를 표시하는 행위와 객체에 조작(`print`) 하는 것은 순서에 상관이 없다. 다음에 코드 셀을 수행한 다음, `display(output)`의 주석을 없애고 다시 수행하며 비교 하자.

````python
import ipywidgets as widgets
from IPython.display import display

output = widgets.Output()

# 위젯 객체를 표시 합니다.
# display(output)

with output:
    print("Hello, World!")
````

- ipywidgets.Button

    Button 위젯은 사용자가 클릭할 수 있는 버튼을 화면에 만들고, 버튼이 클릭되었을 때 특정 함수가 실행되도록 연결할 수 있는 GUI 구성 요소

````python
import ipywidgets as widgets
from IPython.display import display

button = widgets.Button(description="클릭하세요")
output = widgets.Output()

def button_clicked(b):
    with output:
        print("버튼이 클릭되었습니다.")

# 버튼 클릭 이벤트 핸들러를 등록 합니다.
button.on_click(button_clicked)

# 위젯 객체를 표시 합니다.
display(button, output)
````

터미널 화면을 지우고자 할 때 `cls`(MS-DOS) 또는 `clear`(Linux/Unix) 명령을 사용한다. Output 위젯도 출력 내용을 지울 수 있다.

````python
import ipywidgets as widgets
from IPython.display import display

button = widgets.Button(description="클릭하세요")
output = widgets.Output()

click_count = 0

def button_clicked(b):
    global click_count
    click_count += 1
    
    with output:
        output.clear_output()  # 이전 출력 내용을 지웁니다.
        print(f"버튼이 {click_count}번 클릭되었습니다.")

# 버튼 클릭 이벤트 핸들러를 등록 합니다.
button.on_click(button_clicked)

# 위젯 객체를 표시 합니다.
display(button, output)
````

- ipywidgets.Text

     Text 위젯은 사용자가 입력할 수 있는 입력 상자를 화면에 만드는 GUI 구성 요소이다. Text 와 IntText, FloatText 을 선택 할 수 있다. 

````python
import ipywidgets as widgets
from IPython.display import display

num_1 = widgets.IntText(description="숫자 1:")
num_2 = widgets.IntText(description="숫자 2:")
button = widgets.Button(description="더하기")
output = widgets.Output()

def button_clicked(b):
    with output:
        output.clear_output()
        result = num_1.value + num_2.value
        print(f"{num_1.value} + {num_2.value} = {result}")


# 버튼 클릭 이벤트 핸들러를 등록 합니다.
button.on_click(button_clicked)

# 위젯 객체를 표시 합니다.
display(num_1, num_2, button, output)
````

- ipywidgets.Dropdown

````python
import ipywidgets as widgets
from IPython.display import display

dropdown = widgets.Dropdown(
    options=['옵션 1', '옵션 2', '옵션 3'],
    description='선택:',
)
output = widgets.Output()

def dropdown_changed(change):
    with output:
        output.clear_output()
        print(f"선택된 옵션: {change['new']}")

# 드롭다운 메뉴 변경 이벤트 핸들러를 등록 합니다.
dropdown.observe(dropdown_changed, names='value')   

# 위젯 객체를 표시 합니다.
display(dropdown, output)
````

- ipywidgets.RadioButtons

````python
import ipywidgets as widgets
from IPython.display import display

radio = widgets.RadioButtons(
    options=['옵션 A', '옵션 B', '옵션 C'],
    description='선택:',
)
output = widgets.Output()

def radio_changed(change):
    with output:
        output.clear_output()
        print(f"선택된 옵션: {change['new']}")

# 라디오 버튼 변경 이벤트 핸들러를 등록 합니다.
radio.observe(radio_changed, names='value')   

# 위젯 객체를 표시 합니다.
display(radio, output)
````

- ipywidgets.Checkbox

````python
import ipywidgets as widgets
from IPython.display import display

checkbox = widgets.Checkbox(
    description='옵션 A',
)
output = widgets.Output()

def checkbox_changed(change):
    with output:
        output.clear_output()
        print(f"옵션 A: {change['new']}")

# 체크박스 변경 이벤트 핸들러를 등록 합니다.
checkbox.observe(checkbox_changed, names='value')   

# 위젯 객체를 표시 합니다.
display(checkbox, output)
````
