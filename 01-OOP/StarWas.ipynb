# 객체 지향 프로그래밍 (Object-Oriented Programming)

* 이름을 있으면 모두 객체이다.
* 객체는 데이터와 동작으로 정의(표현/설명) 할 수 있다.
* 객체를 정의하는 것은 설계도를 작성하는 것으로 OOP 에서는 클래스(class)라고 부른다.
* 객체를 생성하는 것은 설계도에 따라 실체를 생성(제품화)하는 것으로 OOP 에서는 인스턴스(instance)라고 부른다.
