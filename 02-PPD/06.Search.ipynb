## 순차검색

### 존재 유무 확인

가장 직관적인 검색 방법은 리스트 맨 앞에서 시작하여 순서대로 검사하는 것이다. 검사하다가 키를 발견하면 즉시 바로 True 값을 리턴하면서 검색을 끝낸다. 끝가지 키를 모두 검사해서 찾지 못한 경우에만 False값을 리턴한다. 이 알고리즘을 순차검색(sequential search)이라 하며, 리스트의 귀납구조를 기반으로 반복조거노가 종료조건으로 나눠 다음과 같이 재귀 함수로 작성할 수 있다.

- 재귀 함수 버전

````python
def seq_search_OX(s,x):
    if s != []:
        if s[0] == x:
            return True
        else:
            return seq_search_OX(s[1:],x)
    else:
        return False
    
# Test code
seq_search_OX([3, 5, 4, 2], 4)
````

- while 루프 버전

````python
def seq_search_OX(s,x):
    while s != []:
        if s[0] == x:
            return True
        else:
            s = s[1:]
    return False

# Test code
seq_search_OX([3, 5, 4, 2], 4)
````

- for 루프 버전

````python
def seq_search_OX(s,x):
    for key in s:
        if key == x:
            return True
    return False

# Test code
seq_search_OX([3, 5, 4, 2], 4)
````

- 논리 연산자로 흐름 제어

````python
def seq_search_OX(s,x):
    return s != [] and (s[0] == x or seq_search_OX(s[1:],x))

# Test code
seq_search_OX([3, 5, 4, 2], 4)
````

## 이분검색

### 존재 유무 확인

리스트가 정렬되어 있다면, 리스트 전체를 다 검사해 보지 않고도 키의 존재유무를 판단할 수 있다. 정렬된 리스트 ss를 반으로 나눠 가운데 원소와 키 x를 비교하면, 좌우 반쪽 중 어느 쪽에 검색 키가 있을지 판단 할 수 있다. 리스트 구조를 기반으로 반복조건과 종료조건을 나누어 알고리즘을 작성할 수 있다. 

- 이분검색

````python
def bin_search_OX(ss,x):
    if ss != []:
        mid = len(ss) // 2
        if x == ss[mid]:
            return True
        elif x < ss[mid]:
            return bin_search_OX(ss[:mid],x)
        else:
            return bin_search_OX(ss[mid+1:],x)
    else:
        return False

# Test code
s = [3,5,8,7,4,6,1,9,2]
s.sort()
print(bin_search_OX(s,5))
print(bin_search_OX(s,8))
print(bin_search_OX(s,1))
print(bin_search_OX(s,11))
````

### 위치 인덱스 찾기

검색 범위를 좁혀가면서 검색 대상이 되는 부분의 오리지널 인덱스를 유지하려면 어떻게 해야 할까?  
검색 대상 부분의 시작과 끝을 오리지널 인덱스로 기억할 변수 low와 high를 추가한다. 검색할 반쪽이 정히질 때마다. 그 부분의 시작과 끝에 해당하는 오리지널 인덱스를 각각 계산하여 low와 high 변수에 기억한다. 

- low와 high 변수의 초기 값 설정:
    ```text
    low = 0
    high = len(ss) - 1
    ```

- 반복 조건:
    ```text
    low <= high
    ```

- 종료 조건:
    ```text
    low > high
    ```

````python
def bin_search(ss,x):
    low = 0
    high = len(ss) - 1
    while low <= high:
        mid = (high + low) // 2
        if x == ss[mid]:
            return mid
        elif x < ss[mid]:
            high = mid - 1
        else:
            low = mid + 1
    return None


# Test code
s = [3,5,8,7,4,6,1,9,2]
s.sort()
print(bin_search(s,5))
print(bin_search(s,8))
print(bin_search(s,1))
print(bin_search(s,11))
````

### 함수 호출의 실행 시간 측정 방법

```text
time.perf_counter()
```

````python
import time

print("Now, go!")
start = time.perf_counter()
print("Start at", start)

time.sleep(5)

finish = time.perf_counter()
print("Finish at", finish)

print("Total =", finish - start, "seconds")
````

0 부터 9,999,999 사이의 정수로 구성된 1천만(=10,000,000) 개의 키 모집단에서 키를 8백만(=8,000,000) 개 무작위로 뽑은 검색 대상 리스트는 
```text
random.sample(range(10000000), 8000000)
```
으로 만들 수 있다.

````python
from random import sample, randrange
from time import perf_counter

print("Preparing data for seq_search. Please, wait a moment ...")
data = sample(range(10000000),8000000)

print("Testing seq_search function begins ...")
for _ in range(10):
    x = randrange(10000000)
    start = perf_counter()
    index = seq_search(data, x)
    finish = perf_counter()
    print(x, "is found at", index, "in", finish - start, "seconds")

print("Preparing data for bin_search. Please, wait a moment ...")
data.sort()

print("Testing bin_search function begins ...")
for _ in range(10):
    x = randrange(10000000)
    start = perf_counter()
    index = bin_search(data, x)
    finish = perf_counter()
    print(x, "is found at", index, "in", finish - start, "seconds")
````
